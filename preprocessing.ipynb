```python
def data_quality_report(df, name):
    report = pd.DataFrame({
        "column": df.columns,
        "dtype": df.dtypes.astype(str).values,
        "missing_count": df.isna().sum().values,
        "missing_%": (df.isna().mean() * 100).round(2).values,
        "unique_values": df.nunique(dropna=True).values
    })

    print(f"\n{'='*60}")
    print(f"{name.upper()} DATA QUALITY REPORT")
    print(f"{'='*60}")
    print(f"Rows: {len(df):,}")
    print(f"Columns: {len(df.columns)}")
    print(f"Duplicate rows: {df.duplicated().sum():,}")

    display(report)


data_quality_report(transactions, "Transactions")
data_quality_report(kyc, "KYC")
data_quality_report(merchants, "Merchants")
data_quality_report(chargebacks, "Chargebacks")
```

```python
def clean_timestamp(x):
    if pd.isna(x):
        return pd.NaT

    x = str(x).strip()

    # Unix timestamp
    if x.isdigit() and len(x) >= 9:
        return pd.to_datetime(
            int(x),
            unit="s",
            errors="coerce"
        )

    # Normal datetime formats
    formats = [
        "%Y-%m-%d %H:%M:%S",
        "%m-%d-%Y %I:%M:%S %p",
        "%d/%m/%Y %H:%M:%S",
        "%d-%m-%Y %H:%M:%S",
        "%Y/%m/%d",
    ]

    for fmt in formats:
        try:
            return pd.to_datetime(x, format=fmt)
        except:
            pass

    return pd.NaT
```

```python
def clean_dob(x):
    if pd.isna(x):
        return pd.NaT

    x = str(x).strip()

    # Unix timestamp, including positive and negative values
    if x.lstrip("-").isdigit():
        return pd.to_datetime(
            int(x),
            unit="s",
            errors="coerce"
        )

    formats = [
        "%Y-%m-%d %H:%M:%S",
        "%Y-%m-%d",
        "%Y/%m/%d",
        "%d/%m/%Y %I:%M %p",
        "%d/%m/%Y",
        "%m-%d-%Y %I:%M:%S %p",
        "%m-%d-%Y",
        "%d-%b-%Y",
    ]

    for fmt in formats:
        try:
            return pd.to_datetime(x, format=fmt)
        except:
            pass

    return pd.NaT
```

```python
def clean_income(x):
    if pd.isna(x):
        return np.nan

    x = str(x).strip().upper()

    # Remove currency symbols/text
    x = x.replace("₹", "")
    x = x.replace("INR", "")
    x = x.replace("RS.", "")
    x = x.replace("RS", "")

    # Remove commas
    x = x.replace(",", "")

    # Handle values like 27.3K
    if x.endswith("K"):
        try:
            return float(x[:-1]) * 1000
        except:
            return np.nan

    return pd.to_numeric(x, errors="coerce")
```

```python
def clean_signup_timestamp(x):
    if pd.isna(x):
        return pd.NaT

    x = str(x).strip()

    # Handle Unix timestamp, including negative values
    if x.lstrip("-").isdigit() and len(x.lstrip("-")) >= 9:
        return pd.to_datetime(
            int(x),
            unit="s",
            errors="coerce"
        )

    formats = [
        "%Y-%m-%d %H:%M:%S",
        "%Y-%m-%d",

        "%d/%m/%Y %I:%M %p",
        "%d/%m/%Y",

        "%d-%m-%Y %I:%M %p",
        "%d-%m-%Y",

        "%m-%d-%Y %I:%M %p",
        "%m-%d-%Y",

        "%Y/%m/%d %H:%M:%S",
        "%Y/%m/%d",

        "%d-%b-%Y %H:%M:%S",
        "%d-%b-%Y",

        "%d-%B-%Y %H:%M:%S",
        "%d-%B-%Y",
    ]

    for fmt in formats:
        try:
            return pd.to_datetime(x, format=fmt)
        except:
            pass

    return pd.NaT
```

```python
def clean_business_type(x):
    if pd.isna(x):
        return "UNKNOWN"

    x = str(x).strip().upper().replace("-", "_").replace(" ", "_")

    if x == "INDIVIDUAL":
        return "Individual"

    elif x == "PARTNERSHIP":
        return "Partnership"

    elif x == "SOLE_PROPRIETOR":
        return "Sole Proprietor"

    elif x == "PRIVATE_LIMITED":
        return "Private Limited"

    else:
        return "UNKNOWN"
```

```python
def clean_onboarding_date(x):
    if pd.isna(x):
        return pd.NaT

    x = str(x).strip()

    if x.isdigit() and len(x) >= 9:
        return pd.to_datetime(
            int(x),
            unit="s",
            errors="coerce"
        )

    formats = [
        "%Y-%m-%d %H:%M:%S",
        "%Y-%m-%d",
        "%Y/%m/%d",
        "%d/%m/%Y %I:%M %p",
        "%d/%m/%Y",
        "%m-%d-%Y",
        "%d-%m-%Y",
        "%d-%b-%Y",
        "%d-%B-%Y",
    ]

    for fmt in formats:
        try:
            return pd.to_datetime(x, format=fmt)
        except:
            pass

    return pd.NaT
```

```python
chargebacks["txn_match_flag"] = chargebacks["txn_id_clean"].isin(transactions_clean["txn_id_clean"]).astype(int)
chargebacks["user_match_flag"] = chargebacks["user_id_core"].isin(kyc_master["user_id_core"]).astype(int)
chargebacks["merchant_match_flag"] = chargebacks["merchant_id_core"].isin(merchant_master["merchant_id_core"]).astype(int)

print("Transaction matches:", chargebacks["txn_match_flag"].sum())
print("User matches:", chargebacks["user_match_flag"].sum())
print("Merchant matches:", chargebacks["merchant_match_flag"].sum())
```

## 7. Build chargeback aggregates

The transaction table must remain one row per transaction. We aggregate multiple complaints before joining them so transaction counts and amounts are not inflated.

```python
chargeback_txn = (
    chargebacks
    .groupby("txn_id_clean", dropna=False)
    .agg(
        chargeback_count=("complaint_id_clean", "nunique"),
        disputed_amount=("disputed_amount_calc", "sum"),
        max_severity=("severity_clean", lambda x: ", ".join(sorted(set(x)))),
        delayed_dispute_count=("delayed_dispute_7d_flag", "sum"),
        reporting_delay_days=("reporting_delay_days", "mean")
    )
    .reset_index()
)

chargeback_txn["has_chargeback_flag"] = (chargeback_txn["chargeback_count"] > 0).astype(int)
chargeback_txn["disputed_amount"] = chargeback_txn["disputed_amount"].fillna(0)

print("Chargeback transaction aggregates:", len(chargeback_txn))
```

## 8. Final transaction analytical dataset

```python
analytics_transactions = transactions_clean.copy()

kyc_join = kyc_master.drop_duplicates("user_id_core").copy()
merchant_join = merchant_master.drop_duplicates("merchant_id_core").copy()

kyc_cols = [
    "user_id_core", "kyc_status_clean", "risk_segment_clean",
    "monthly_income_clean", "missing_pan_flag", "missing_aadhaar_flag",
    "missing_dob_flag", "identity_conflict_score"
]

merchant_cols = [
    "merchant_id_core", "merchant_name", "merchant_category_clean",
    "business_type_clean", "merchant_status", "onboarding_date_clean",
    "declared_avg_ticket_clean", "merchant_conflict_score"
]

analytics_transactions = analytics_transactions.merge(
    kyc_join[kyc_cols], on="user_id_core", how="left", validate="many_to_one"
)

analytics_transactions = analytics_transactions.merge(
    merchant_join[merchant_cols], on="merchant_id_core", how="left", validate="many_to_one"
)

analytics_transactions = analytics_transactions.merge(
    chargeback_txn, on="txn_id_clean", how="left", validate="one_to_one"
)

analytics_transactions["chargeback_count"] = analytics_transactions["chargeback_count"].fillna(0)
analytics_transactions["disputed_amount"] = analytics_transactions["disputed_amount"].fillna(0)
analytics_transactions["has_chargeback_flag"] = analytics_transactions["has_chargeback_flag"].fillna(0).astype(int)
analytics_transactions["delayed_dispute_count"] = analytics_transactions["delayed_dispute_count"].fillna(0)

print("Transaction rows before joins:", len(transactions_clean))
print("Analytics transaction rows:", len(analytics_transactions))
print("Duplicate transaction IDs:", analytics_transactions["txn_id_clean"].duplicated().sum())
```

## 9. Add time and behavioral features

```python
analytics_transactions["date"] = analytics_transactions["timestamp_clean"].dt.date
analytics_transactions["hour"] = analytics_transactions["timestamp_clean"].dt.hour
analytics_transactions["day_of_week"] = analytics_transactions["timestamp_clean"].dt.day_name()
analytics_transactions["is_weekend"] = (analytics_transactions["timestamp_clean"].dt.dayofweek >= 5).astype(int)

user_stats = analytics_transactions.groupby("user_id_core")["amount_clean"].agg(
    user_txn_count="count",
    user_avg_amount="mean",
    user_total_amount="sum",
    user_std_amount="std"
).reset_index()

merchant_stats = analytics_transactions.groupby("merchant_id_core")["amount_clean"].agg(
    merchant_txn_count="count",
    merchant_avg_amount="mean",
    merchant_total_amount="sum",
    merchant_std_amount="std"
).reset_index()

analytics_transactions = analytics_transactions.merge(user_stats, on="user_id_core", how="left")
analytics_transactions = analytics_transactions.merge(merchant_stats, on="merchant_id_core", how="left")

analytics_transactions["user_amount_deviation"] = (
    analytics_transactions["amount_clean"] - analytics_transactions["user_avg_amount"]
) / analytics_transactions["user_std_amount"].replace(0, np.nan)

analytics_transactions["merchant_amount_deviation"] = (
    analytics_transactions["amount_clean"] - analytics_transactions["merchant_avg_amount"]
) / analytics_transactions["merchant_std_amount"].replace(0, np.nan)

analytics_transactions["large_user_deviation_flag"] = (
    analytics_transactions["user_amount_deviation"].abs() >= 3
).fillna(False).astype(int)

analytics_transactions["large_merchant_deviation_flag"] = (
    analytics_transactions["merchant_amount_deviation"].abs() >= 3
).fillna(False).astype(int)
```

## 10. User-level analytics and risk features

```python
user_risk = (
    analytics_transactions
    .groupby("user_id_core")
    .agg(
        transaction_count=("txn_id_clean", "nunique"),
        total_amount=("amount_clean", "sum"),
        average_amount=("amount_clean", "mean"),
        failed_count=("status_clean", lambda x: (x == "FAILED").sum()),
        pending_count=("status_clean", lambda x: (x == "PENDING").sum()),
        unique_merchants=("merchant_id_core", "nunique"),
        chargeback_count=("chargeback_count", "sum"),
        disputed_amount=("disputed_amount", "sum"),
        missing_utr_count=("missing_utr_flag", "sum"),
        negative_amount_count=("negative_amount_flag", "sum"),
        anomalous_amount_count=("large_user_deviation_flag", "sum")
    )
    .reset_index()
)

user_risk["failed_rate"] = user_risk["failed_count"] / user_risk["transaction_count"].replace(0, np.nan)
user_risk["chargeback_ratio"] = user_risk["chargeback_count"] / user_risk["transaction_count"].replace(0, np.nan)

user_profile = kyc_join[[
    "user_id_core", "kyc_status_clean", "risk_segment_clean",
    "missing_pan_flag", "missing_aadhaar_flag", "missing_dob_flag",
    "identity_conflict_score"
]].copy()

user_risk = user_risk.merge(user_profile, on="user_id_core", how="left")

user_risk["risk_score"] = 0.0
user_risk["risk_score"] += user_risk["chargeback_ratio"].fillna(0).clip(0, 1) * 30
user_risk["risk_score"] += user_risk["failed_rate"].fillna(0).clip(0, 1) * 15
user_risk["risk_score"] += (user_risk["identity_conflict_score"].fillna(0).clip(0, 4) / 4) * 20
user_risk["risk_score"] += user_risk["missing_utr_count"].gt(0).astype(int) * 5
user_risk["risk_score"] += user_risk["negative_amount_count"].gt(0).astype(int) * 5
user_risk["risk_score"] += user_risk["anomalous_amount_count"].gt(0).astype(int) * 10
user_risk["risk_score"] += user_risk["kyc_status_clean"].isin(["REJECTED", "PENDING"]).astype(int) * 10
user_risk["risk_score"] += user_risk["risk_segment_clean"].eq("HIGH").astype(int) * 5
user_risk["risk_score"] = user_risk["risk_score"].clip(0, 100).round(2)

user_risk["risk_level"] = pd.cut(
    user_risk["risk_score"],
    bins=[-1, 24.99, 49.99, 74.99, 100],
    labels=["LOW", "MEDIUM", "HIGH", "CRITICAL"]
).astype(str)
```

## 11. Merchant-level analytics and risk features

```python
merchant_risk = (
    analytics_transactions
    .groupby("merchant_id_core")
    .agg(
        transaction_count=("txn_id_clean", "nunique"),
        total_amount=("amount_clean", "sum"),
        average_amount=("amount_clean", "mean"),
        unique_users=("user_id_core", "nunique"),
        failed_count=("status_clean", lambda x: (x == "FAILED").sum()),
        chargeback_count=("chargeback_count", "sum"),
        disputed_amount=("disputed_amount", "sum"),
        delayed_dispute_count=("delayed_dispute_count", "sum"),
        anomalous_amount_count=("large_merchant_deviation_flag", "sum")
    )
    .reset_index()
)

merchant_risk["failed_rate"] = merchant_risk["failed_count"] / merchant_risk["transaction_count"].replace(0, np.nan)
merchant_risk["chargeback_ratio"] = merchant_risk["chargeback_count"] / merchant_risk["transaction_count"].replace(0, np.nan)

merchant_profile = merchant_join[[
    "merchant_id_core", "merchant_name", "merchant_category_clean",
    "business_type_clean", "merchant_status", "merchant_conflict_score"
]].copy()

merchant_risk = merchant_risk.merge(merchant_profile, on="merchant_id_core", how="left")

merchant_risk["risk_score"] = 0.0
merchant_risk["risk_score"] += merchant_risk["chargeback_ratio"].fillna(0).clip(0, 1) * 45
merchant_risk["risk_score"] += merchant_risk["failed_rate"].fillna(0).clip(0, 1) * 15
merchant_risk["risk_score"] += merchant_risk["delayed_dispute_count"].gt(0).astype(int) * 10
merchant_risk["risk_score"] += merchant_risk["anomalous_amount_count"].gt(0).astype(int) * 15
merchant_risk["risk_score"] += (merchant_risk["merchant_conflict_score"].fillna(0).clip(0, 3) / 3) * 15
merchant_risk["risk_score"] = merchant_risk["risk_score"].clip(0, 100).round(2)

merchant_risk["risk_level"] = pd.cut(
    merchant_risk["risk_score"],
    bins=[-1, 24.99, 49.99, 74.99, 100],
    labels=["LOW", "MEDIUM", "HIGH", "CRITICAL"]
).astype(str)
```

## 12. Transaction-level risk score

```python
analytics_transactions = analytics_transactions.merge(
    user_risk[["user_id_core", "risk_score", "risk_level"]].rename(
        columns={"risk_score": "user_risk_score", "risk_level": "user_risk_level"}
    ),
    on="user_id_core", how="left", validate="many_to_one"
)

analytics_transactions = analytics_transactions.merge(
    merchant_risk[["merchant_id_core", "risk_score", "risk_level"]].rename(
        columns={"risk_score": "merchant_risk_score", "risk_level": "merchant_risk_level"}
    ),
    on="merchant_id_core", how="left", validate="many_to_one"
)

analytics_transactions["transaction_risk_score"] = 0.0
analytics_transactions["transaction_risk_score"] += analytics_transactions["user_risk_score"].fillna(0) * 0.30
analytics_transactions["transaction_risk_score"] += analytics_transactions["merchant_risk_score"].fillna(0) * 0.30
analytics_transactions["transaction_risk_score"] += analytics_transactions["large_user_deviation_flag"] * 15
analytics_transactions["transaction_risk_score"] += analytics_transactions["large_merchant_deviation_flag"] * 10
analytics_transactions["transaction_risk_score"] += analytics_transactions["missing_utr_flag"] * 5
analytics_transactions["transaction_risk_score"] += analytics_transactions["negative_amount_flag"] * 10
analytics_transactions["transaction_risk_score"] += analytics_transactions["has_chargeback_flag"] * 20
analytics_transactions["transaction_risk_score"] = analytics_transactions["transaction_risk_score"].clip(0, 100).round(2)

analytics_transactions["transaction_risk_level"] = pd.cut(
    analytics_transactions["transaction_risk_score"],
    bins=[-1, 24.99, 49.99, 74.99, 100],
    labels=["LOW", "MEDIUM", "HIGH", "CRITICAL"]
).astype(str)
```

## 13. Risk explanations

```python
def user_reason(row):
    reasons = []
    if row["chargeback_ratio"] >= 0.1:
        reasons.append("elevated chargeback ratio")
    if row["failed_rate"] >= 0.2:
        reasons.append("high failed transaction rate")
    if row["identity_conflict_score"] >= 2:
        reasons.append("identity conflicts in KYC history")
    if row["kyc_status_clean"] in ["REJECTED", "PENDING"]:
        reasons.append("KYC is not fully approved")
    if row["risk_segment_clean"] == "HIGH":
        reasons.append("high KYC risk segment")
    if row["anomalous_amount_count"] > 0:
        reasons.append("unusual transaction amounts")
    if row["missing_utr_count"] > 0:
        reasons.append("transactions with missing UTR")
    return "; ".join(reasons) if reasons else "No major risk signal"

user_risk["risk_reasons"] = user_risk.apply(user_reason, axis=1)

def merchant_reason(row):
    reasons = []
    if row["chargeback_ratio"] >= 0.1:
        reasons.append("elevated chargeback ratio")
    if row["failed_rate"] >= 0.2:
        reasons.append("high failed transaction rate")
    if row["delayed_dispute_count"] > 0:
        reasons.append("delayed disputes")
    if row["anomalous_amount_count"] > 0:
        reasons.append("unusual transaction amounts")
    if row["merchant_conflict_score"] > 0:
        reasons.append("merchant profile conflicts")
    return "; ".join(reasons) if reasons else "No major risk signal"

merchant_risk["risk_reasons"] = merchant_risk.apply(merchant_reason, axis=1)
```

## 14. Detect suspicious fraud rings

We use a bipartite user-merchant graph. To reduce noise, the graph focuses on users and merchants with elevated risk or chargeback activity. Connected components are then scored using actual transaction and dispute evidence.

```python
graph = nx.Graph()

high_users = set(user_risk.loc[user_risk["risk_score"] >= 50, "user_id_core"].dropna())
high_merchants = set(merchant_risk.loc[merchant_risk["risk_score"] >= 50, "merchant_id_core"].dropna())

suspicious_edges = analytics_transactions[
    analytics_transactions["user_id_core"].notna()
    & analytics_transactions["merchant_id_core"].notna()
    & (
        analytics_transactions["user_id_core"].isin(high_users)
        | analytics_transactions["merchant_id_core"].isin(high_merchants)
        | (analytics_transactions["has_chargeback_flag"] == 1)
    )
].copy()

for _, row in suspicious_edges.iterrows():
    user_node = "U_" + str(row["user_id_core"])
    merchant_node = "M_" + str(row["merchant_id_core"])
    graph.add_edge(user_node, merchant_node)

components = list(nx.connected_components(graph))
print("Graph nodes:", graph.number_of_nodes())
print("Graph edges:", graph.number_of_edges())
print("Connected components:", len(components))
```

## 15. Score fraud rings

```python
ring_rows = []
ring_members = []

for ring_number, component in enumerate(components, start=1):
    users = [x[2:] for x in component if x.startswith("U_")]
    merchants = [x[2:] for x in component if x.startswith("M_")]
    if len(users) < 2 or len(merchants) < 1:
        continue

    ring_txn = suspicious_edges[
        suspicious_edges["user_id_core"].isin(users)
        & suspicious_edges["merchant_id_core"].isin(merchants)
    ]

    total_amount = ring_txn["amount_clean"].sum()
    chargeback_amount = ring_txn["disputed_amount"].sum()
    chargeback_count = ring_txn["chargeback_count"].sum()
    risk_score = min(100, round(
        min(40, len(users) * 5)
        + min(25, len(merchants) * 8)
        + min(25, chargeback_count * 5)
        + min(10, chargeback_amount / max(total_amount, 1) * 100), 2
    ))

    reasons = []
    if len(users) >= 3:
        reasons.append("multiple connected users")
    if len(merchants) >= 2:
        reasons.append("multiple connected merchants")
    if chargeback_count > 0:
        reasons.append("linked chargeback activity")
    if chargeback_amount > 0 and total_amount > 0 and chargeback_amount / total_amount >= 0.05:
        reasons.append("elevated disputed amount ratio")

    ring_id = "RING_" + str(ring_number).zfill(4)
    ring_rows.append({
        "ring_id": ring_id,
        "user_count": len(users),
        "merchant_count": len(merchants),
        "transaction_count": len(ring_txn),
        "total_amount": total_amount,
        "chargeback_count": chargeback_count,
        "chargeback_amount": chargeback_amount,
        "risk_score": risk_score,
        "risk_level": "CRITICAL" if risk_score >= 75 else "HIGH" if risk_score >= 50 else "MEDIUM",
        "risk_reasons": "; ".join(reasons) if reasons else "connected high-risk activity"
    })

    for user in users:
        ring_members.append({"ring_id": ring_id, "member_type": "USER", "member_id": user})
    for merchant in merchants:
        ring_members.append({"ring_id": ring_id, "member_type": "MERCHANT", "member_id": merchant})

fraud_rings = pd.DataFrame(ring_rows).sort_values("risk_score", ascending=False).reset_index(drop=True)
ring_members = pd.DataFrame(ring_members)

print("Suspicious rings:", len(fraud_rings))
display(fraud_rings.head(10))
```

```python
app_code = r'''import streamlit as st
import pandas as pd
import numpy as np
from pathlib import Path
import plotly.express as px

st.set_page_config(page_title="FraudLens AI", page_icon="🛡️", layout="wide")

BASE = Path(__file__).resolve().parent
DATA = BASE / "fraudlens_output"

@st.cache_data
def load_data():
    files = {
        "tx": "analytics_transactions.csv",
        "users": "user_risk.csv",
        "merchants": "merchant_risk.csv",
        "chargebacks": "chargebacks_clean.csv",
        "rings": "fraud_rings.csv",
        "members": "ring_members.csv",
        "daily": "daily_metrics.csv",
        "kyc": "kyc_master.csv",
        "summary": "summary.csv"
    }
    return {k: pd.read_csv(DATA / v) for k, v in files.items()}

d = load_data()
tx = d["tx"]
users = d["users"]
merchants = d["merchants"]
chargebacks = d["chargebacks"]
rings = d["rings"]
daily = d["daily"]
summary = d["summary"]

for frame in [tx, users, merchants, chargebacks, rings, daily]:
    for col in frame.columns:
        if frame[col].dtype == "object":
            frame[col] = frame[col].fillna("UNKNOWN")

st.title("🛡️ FraudLens AI")
st.caption("UPI Fraud Ring Detection & Merchant Risk Intelligence")

pages = ["Executive Overview", "Transactions", "Merchant Intelligence", "User & KYC Risk", "Chargebacks", "Fraud Rings", "AI Investigator"]
page = st.sidebar.radio("Navigate", pages)

st.sidebar.markdown("### Filters")
if "merchant_category_clean" in tx.columns:
    cats = sorted(tx["merchant_category_clean"].dropna().astype(str).unique())
    selected_cats = st.sidebar.multiselect("Merchant category", cats, default=cats)
else:
    selected_cats = []

if "transaction_risk_level" in tx.columns:
    levels = ["LOW", "MEDIUM", "HIGH", "CRITICAL"]
    selected_levels = st.sidebar.multiselect("Transaction risk", levels, default=levels)
else:
    selected_levels = []

filtered_tx = tx.copy()
if selected_cats:
    filtered_tx = filtered_tx[filtered_tx["merchant_category_clean"].astype(str).isin(selected_cats)]
if selected_levels:
    filtered_tx = filtered_tx[filtered_tx["transaction_risk_level"].astype(str).isin(selected_levels)]


def money(x):
    return f"₹{x/1e7:.2f} Cr" if abs(x) >= 1e7 else f"₹{x/1e5:.2f} L"


def pct(x):
    return f"{x*100:.2f}%"

if page == "Executive Overview":
    total_tx = filtered_tx["txn_id_clean"].nunique()
    total_amount = filtered_tx["amount_clean"].sum()
    cb_count = filtered_tx["chargeback_count"].sum()
    disputed = filtered_tx["disputed_amount"].sum()
    high_tx = filtered_tx["transaction_risk_level"].isin(["HIGH", "CRITICAL"]).sum()
    failed = (filtered_tx["status_clean"] == "FAILED").mean()

    a,b,c,e,f,g = st.columns(6)
    a.metric("Transactions", f"{total_tx:,}")
    b.metric("Transaction Value", money(total_amount))
    c.metric("Chargebacks", f"{int(cb_count):,}")
    e.metric("Disputed Amount", money(disputed))
    f.metric("High/Critical Txns", f"{int(high_tx):,}")
    g.metric("Failed Rate", pct(failed))

    left, right = st.columns(2)
    with left:
        chart = filtered_tx.copy()
        chart["date"] = pd.to_datetime(chart["timestamp_clean"]).dt.date
        trend = chart.groupby("date", as_index=False)["amount_clean"].sum()
        st.plotly_chart(px.line(trend, x="date", y="amount_clean", title="Transaction Value Trend"), use_container_width=True)
    with right:
        risk = filtered_tx["transaction_risk_level"].value_counts().reindex(["LOW","MEDIUM","HIGH","CRITICAL"], fill_value=0).reset_index()
        risk.columns = ["risk_level", "count"]
        st.plotly_chart(px.bar(risk, x="risk_level", y="count", title="Transaction Risk Distribution"), use_container_width=True)

    st.subheader("Top Risk Signals")
    st.dataframe(users.sort_values("risk_score", ascending=False)[[c for c in ["user_id_core","risk_score","risk_level","chargeback_ratio","failed_rate","risk_reasons"] if c in users.columns]].head(10), use_container_width=True)

elif page == "Transactions":
    st.header("Transaction Intelligence")
    st.write(f"Showing {len(filtered_tx):,} transactions after filters.")
    cols = [c for c in ["txn_id_clean","timestamp_clean","user_id_core","merchant_id_core","amount_clean","status_clean","chargeback_count","disputed_amount","transaction_risk_score","transaction_risk_level"] if c in filtered_tx.columns]
    st.dataframe(filtered_tx[cols].sort_values("transaction_risk_score", ascending=False).head(500), use_container_width=True)

    trend = filtered_tx.copy()
    trend["date"] = pd.to_datetime(trend["timestamp_clean"]).dt.date
    daily_tx = trend.groupby("date", as_index=False).agg(transaction_count=("txn_id_clean","nunique"), transaction_amount=("amount_clean","sum"), failed_count=("status_clean", lambda x: (x=="FAILED").sum()))
    st.plotly_chart(px.line(daily_tx, x="date", y="transaction_count", title="Daily Transaction Volume"), use_container_width=True)
    st.plotly_chart(px.line(daily_tx, x="date", y="failed_count", title="Daily Failed Transactions"), use_container_width=True)

elif page == "Merchant Intelligence":
    st.header("Merchant Risk Intelligence")
    top = merchants.sort_values("risk_score", ascending=False).copy()
    cols = [c for c in ["merchant_id_core","merchant_name","merchant_category_clean","transaction_count","total_amount","chargeback_count","chargeback_ratio","failed_rate","risk_score","risk_level","risk_reasons"] if c in top.columns]
    st.dataframe(top[cols].head(50), use_container_width=True)
    st.plotly_chart(px.bar(top.head(15), x="merchant_name" if "merchant_name" in top.columns else "merchant_id_core", y="risk_score", title="Top Merchant Risk Scores"), use_container_width=True)
    cat = merchants.groupby("merchant_category_clean", dropna=False).agg(transaction_count=("transaction_count","sum"), disputed_amount=("disputed_amount","sum"), chargeback_count=("chargeback_count","sum")).reset_index().sort_values("disputed_amount", ascending=False)
    st.plotly_chart(px.bar(cat, x="merchant_category_clean", y="disputed_amount", title="Disputed Amount by Merchant Category"), use_container_width=True)

elif page == "User & KYC Risk":
    st.header("User & KYC Risk")
    risk_cols = [c for c in ["user_id_core","transaction_count","total_amount","chargeback_count","disputed_amount","chargeback_ratio","failed_rate","identity_conflict_score","kyc_status_clean","risk_segment_clean","risk_score","risk_level","risk_reasons"] if c in users.columns]
    st.dataframe(users.sort_values("risk_score", ascending=False)[risk_cols].head(100), use_container_width=True)
    if "kyc_status_clean" in users.columns:
        kyc_dist = users["kyc_status_clean"].value_counts().reset_index()
        kyc_dist.columns = ["status", "count"]
        st.plotly_chart(px.pie(kyc_dist, names="status", values="count", title="KYC Status Distribution"), use_container_width=True)

elif page == "Chargebacks":
    st.header("Chargeback Intelligence")
    cb_amount = chargebacks["disputed_amount_calc"].sum() if "disputed_amount_calc" in chargebacks.columns else chargebacks.get("disputed_amount_clean", pd.Series(dtype=float)).sum()
    a,b,c = st.columns(3)
    a.metric("Chargeback Records", f"{len(chargebacks):,}")
    b.metric("Disputed Amount", money(cb_amount))
    c.metric("Delayed > 7 Days", f"{int(chargebacks.get('delayed_dispute_7d_flag', pd.Series(dtype=int)).sum()):,}")
    if "reason_code_clean" in chargebacks.columns:
        reason = chargebacks["reason_code_clean"].value_counts().reset_index()
        reason.columns = ["reason", "count"]
        st.plotly_chart(px.bar(reason, x="reason", y="count", title="Chargeback Reason Distribution"), use_container_width=True)
    st.dataframe(chargebacks.head(500), use_container_width=True)

elif page == "Fraud Rings":
    st.header("Suspicious Fraud Rings")
    st.caption("A ring is treated as suspicious connected activity supported by risk and/or dispute evidence; it is not a proof of criminal conduct.")
    if len(rings):
        st.dataframe(rings.sort_values("risk_score", ascending=False).head(100), use_container_width=True)
        selected = st.selectbox("Inspect ring", rings["ring_id"].astype(str).tolist())
        ring = rings[rings["ring_id"].astype(str) == selected].iloc[0]
        members = d["members"][d["members"]["ring_id"].astype(str) == selected]
        a,b,c,d1 = st.columns(4)
        a.metric("Risk Score", f"{ring['risk_score']:.1f}")
        b.metric("Users", int(ring["user_count"]))
        c.metric("Merchants", int(ring["merchant_count"]))
        d1.metric("Chargebacks", int(ring["chargeback_count"]))
        st.write("**Evidence:**", ring.get("risk_reasons", ""))
        st.dataframe(members, use_container_width=True)
    else:
        st.info("No suspicious rings were produced by the current graph criteria.")

elif page == "AI Investigator":
    st.header("🤖 AI Investigator")
    st.caption("Grounded investigator: answers are generated from the exported FraudLens analytics tables.")
    query = st.text_input("Ask an investigation question", placeholder="Which merchant has the highest chargeback ratio?")
    q = query.lower().strip()
    if q:
        if "highest chargeback ratio" in q or ("merchant" in q and "chargeback ratio" in q):
            row = merchants.sort_values("chargeback_ratio", ascending=False).iloc[0]
            st.success(f"Top merchant by chargeback ratio: {row.get('merchant_name', row['merchant_id_core'])} with {row['chargeback_ratio']*100:.2f}%.")
            st.dataframe(pd.DataFrame([row]))
        elif "highest disputed" in q and "merchant" in q:
            row = merchants.sort_values("disputed_amount", ascending=False).iloc[0]
            st.success(f"Merchant with highest disputed amount: {row.get('merchant_name', row['merchant_id_core'])} with {money(row['disputed_amount'])}.")
            st.dataframe(pd.DataFrame([row]))
        elif "highest disputed" in q and "user" in q:
            row = users.sort_values("disputed_amount", ascending=False).iloc[0]
            st.success(f"User with highest disputed amount: {row['user_id_core']} with {money(row['disputed_amount'])}.")
            st.dataframe(pd.DataFrame([row]))
        elif "daily" in q and "volume" in q:
            st.plotly_chart(px.line(daily, x="date", y="transaction_count", title="Daily Transaction Volume"), use_container_width=True)
        elif "reason" in q and "chargeback" in q:
            reason = chargebacks["reason_code_clean"].value_counts().reset_index()
            reason.columns = ["reason", "count"]
            st.dataframe(reason, use_container_width=True)
        elif "ring" in q:
            st.dataframe(rings.sort_values("risk_score", ascending=False).head(20), use_container_width=True)
        elif "high risk" in q and "user" in q:
            st.dataframe(users[users["risk_level"].isin(["HIGH","CRITICAL"])].sort_values("risk_score", ascending=False).head(50), use_container_width=True)
        else:
            st.info("Try: highest chargeback ratio merchant, highest disputed merchant, highest disputed user, daily transaction volume, chargeback reasons, suspicious rings, or high risk users.")

st.sidebar.caption("FraudLens AI • Datathon decision-support prototype")
'''
app_path = PROJECT_DIR / "app.py"
app_path.write_text(app_code, encoding="utf-8")
print("Streamlit app created:", app_path)
```
